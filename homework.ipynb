```python
# 6
media_original = df['horsepower'].mean()
print("Média original:", media_original)
```

```python
df['horsepower_preenchida'] = df['horsepower'].fillna(media_original)
```

```python
media_nova = df['horsepower_preenchida'].mean()
print("Média nova:", media_nova)
```

```python
# 7

# 
df_asia = df[df['origin'] == 'Asia']
X = df_asia[['engine_displacement', 'horsepower', 'vehicle_weight']].iloc[:7].values

# 
XT = X.T

# 
XTX = XT @ X

# 
XTX_inv = np.linalg.inv(XTX)

# 
print("Resultado final:", XTX_inv.sum())
```
